```python
%%writefile patient_data.py

# Problem Set 4: Functions and Modules
# Exercise 1 and Exercise 2
# patient_data.py

# Store all fictional patient records
patients = [
    ("Alan Reed", "000-00-0001", 74, "04/12/1952", 70, 218, "M"),
    ("Maria Lopez", "000-00-0002", 76, "09/03/1950", 64, 155, "F"),
    ("Robert Chen", "000-00-0003", 68, "01/19/1958", 72, 225, "M"),
    ("James Hill", "000-00-0004", 82, "11/28/1943", 69, 210, "M"),
    ("Susan Patel", "000-00-0005", 72, "06/14/1954", 65, 205, "F"),
    ("David King", "000-00-0006", 71, "02/07/1955", 71, 195, "M"),
    ("Henry Cole", "000-00-0007", 79, "08/21/1947", 68, 242, "M"),
    ("Linda Grant", "000-00-0008", 66, "12/02/1959", 63, 180, "F")
]


# Return the complete patient list
def get_patients():
    """Return all patient records."""
    return patients


# Check if a patient meets all three screening conditions
def screening_rule(patient):
    """Return True if the patient meets the screening rule."""

    # Get age, weight, and gender from the patient tuple
    age = patient[2]
    weight = patient[5]
    gender = patient[6]

    # Return True only if all three conditions are met
    return gender == "M" and age > 70 and weight > 200


# Print one patient's information in a readable format
def display_patient(patient):
    """Display one patient record."""

    print("Name:", patient[0])
    print("SSN:", patient[1])
    print("Age:", patient[2])
    print("DoB:", patient[3])
    print("Height:", patient[4], "in")
    print("Weight:", patient[5], "lb")
    print("Gender:", patient[6])
```

```python
%%writefile heart_screen.py

# Exercise 3: Screening Program
# heart_screen.py

# Import the patient data module
import patient_data

# Get all patient records
patients = patient_data.get_patients()

# Print the heading
print("Patients Matching Screening Rule")
print()

# Check each patient in the list
for patient in patients:

    # Display the patient only if the screening rule is True
    if patient_data.screening_rule(patient):
        patient_data.display_patient(patient)
        print("")
```
