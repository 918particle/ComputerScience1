```python
def add_course(courses, course):
  courses.append(course)

def get_time(course):
  return course[5]

def sort_by_time(courses):
  courses.sort(key=get_time)

def display_courses(courses):
  for course in courses:
    print(course[0],
    course[1],
    course[4],
    f"{course[5]:04d}",
    course[2],
    course[3])

def same_room(c1, c2):
  is_same_building = c1[2] == c2[2]  #T/F for the same building
  is_same_room = c1[3] == c2[3] #T/F for same room
  is_same_time = c1[5] == c2[5] #T/F for the same time
  is_same_day = not set(c1[4]).isdisjoint(c2[4]) #T/F is the letters match
  return is_same_building and is_same_room and is_same_time and is_same_day
```

```python
courses = [
["MATH",100,"SLC",202,"MWF",1200],
["ECON",303,"SLC",101,"MF",1400],
["ENGLISH",400,"SLC",300,"M",1300],
["WRITING",700,"SLC",400,"MWF",1500],
]

#sort_by_time(courses)
#display_courses(courses)
for i in range(len(courses)):
  for j in range(i+1, len(courses)):
    if not same_room(courses[i], courses[j]):
      print(courses[i][0], courses[j][0])
```
