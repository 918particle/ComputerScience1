```python
%%writefile course_tools.py
#Records for meeting days and times
courses = [
    ["PHYS", 100, "SLC", 202, "MTF", 1400],
    ["MATH", 150, "SLC", 105, "MTWF", 1000],
    ["BIOL", 101, "SLC", 210, "MWF", 900],
    ["PHYS", 200, "SLC", 202, "TTh", 1100]
]
#added course
course = ["PHYS" , 100, "SLC", 202, "MWF", 1400]

#add course to end of courses
def add_course(courses, course):
  courses.append(course)

#gets the time from course 5 using return function
def get_time(course):
  return course[5]

#puts the classes in chronological order (time wise)
def sort_by_time(courses):
  courses.sort(key=get_time)

#prints (courses =) and the appended course
def display_courses(courses):
  for course in courses:
    print(course[0],
          course[1],
          course[4],
          f"{course[5]:04d}",
          course[2],
          course[3])

#organizes the courses to see if they conflict
def has_conflict(course_a, course_b):
  same_room = course_a[2] == course_b[2] and course_a[3] == course_b[3]
  same_time = course_a[5] == course_b[5]
  shared_day = any(day in course_b[4] for day in course_a[4])

  return same_room and same_time and shared_day


def find_conflicts(courses):
    conflicts = []
    for i in range(len(courses)):
        for j in range(i + 1, len(courses)):
            if has_conflict(courses[i], courses[j]):
                conflicts.append((courses[i], courses[j]))
    return conflicts

```

```python
import course_tools

courses = []

# My courses this semester
course_tools.add_course(courses, ["PHYS", 100, "SLC", 202, "MWF", 1400])
course_tools.add_course(courses, ["MATH", 210, "COL", 115, "TR", 930])
course_tools.add_course(courses, ["CS", 150, "ENG", 340, "MWF", 1100])
course_tools.add_course(courses, ["ENGL", 101, "HUM", 210, "TR", 1530])

#A "classmates" courses to challenge
course_tools.add_course(courses, ["PHYS", 100, "SLC", 202, "MWF", 1400])
course_tools.add_course(courses, ["HIST", 120, "HUM", 210, "TR", 1530])
course_tools.add_course(courses, ["BIO", 220, "SCI", 118, "MWF", 930])

#Prediction before sorting earliest time in the list is 930 (MATH 210 / BIO 220) Latest time in the list is 1530 (ENGL 101 / HIST 120)

course_tools.sort_by_time(courses)

print("Sorted Schedule")
print("")
course_tools.display_courses(courses)

print("")
print("Room Conflicts")
print("")

conflicts = course_tools.find_conflicts(courses)

if conflicts:
    for course_a, course_b in conflicts:
        print(f"{course_a[0]} {course_a[1]}  <-- conflicts with -->  "
              f"{course_b[0]} {course_b[1]}  "
              f"({course_a[2]} {course_a[3]}, {course_a[4]}, "
              f"{course_a[5]:04d})")
else:
    print("No conflicts found.")
```
