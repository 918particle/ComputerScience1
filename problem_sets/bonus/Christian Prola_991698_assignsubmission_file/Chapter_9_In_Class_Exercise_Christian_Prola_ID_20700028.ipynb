```python
%%writefile course_tools.py

# Chapter 9 In-Class Exercise
# Course Schedules and Modules
# course_tools.py

# Add one course to the course list
def add_course(courses, course):
    courses.append(course)


# Return the meeting time for one course
def get_time(course):
    return course[5]


# Sort all courses by meeting time
def sort_by_time(courses):
    courses.sort(key=get_time)


# Display all courses in a readable format
def display_courses(courses):
    for course in courses:
        print(
            course[0],
            course[1],
            course[4],
            f"{course[5]:04d}",
            course[2],
            course[3]
        )


# Check whether two courses have a room conflict
def room_conflict(course1, course2):

    # Check building, room, and meeting time
    same_building = course1[2] == course2[2]
    same_room = course1[3] == course2[3]
    same_time = course1[5] == course2[5]

    # Check whether they share at least one meeting day
    shared_day = False

    for day in course1[4]:
        if day in course2[4]:
            shared_day = True

    # A conflict requires all conditions to be True
    return same_building and same_room and same_time and shared_day
```

```python
%%writefile course_schedule.py

# Chapter 9 In-Class Exercise
# Course Schedule Program
# course_schedule.py

# Import the course tools module
import course_tools

# Create an empty course list
courses = []

# Add my courses
# Format:
# [subject, number, building, room, days, time]

course_tools.add_course(
    courses,
    ["COSC", 120, "SLC", 416, "MW", 830]
)

course_tools.add_course(
    courses,
    ["MATH", 85, "SLC", 510, "MTWF", 1000]
)

course_tools.add_course(
    courses,
    ["BSAD", 130, "Hoover Memorial Hall", 104, "TR", 1100]
)

course_tools.add_course(
    courses,
    ["POET", 101, "SLC", 205, "M", 1200]
)

course_tools.add_course(
    courses,
    ["POET", 100, "SLC", 311, "MWF", 1330]
)

# A classmate also takes POET 101 and POET 100 with me on Monday.
# These shared courses are already included in the schedule above.

# Sort the courses by meeting time
course_tools.sort_by_time(courses)

# Display the sorted schedule
print("Course Schedule")
course_tools.display_courses(courses)

# Predict the first and last course
print()
print("First course:", courses[0][0], courses[0][1])
print("Last course:", courses[-1][0], courses[-1][1])
```
