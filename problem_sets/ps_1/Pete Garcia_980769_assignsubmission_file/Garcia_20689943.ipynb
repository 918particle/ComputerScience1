```python
from math import cos
print (0.0, cos (0.0))
print (0.5, cos (0.5))
print (1.0, cos (1.0))
print (1.5, cos (1.5))
print (2.0, cos (2.0))
print (2.5, cos (2.5))
print (3.0, cos (3.0))
print (3.5, cos (3.5))
print (4.0, cos (4.0))
print (4.5, cos (4.5))
```

```python
import matplotlib.pyplot as plt
from math import cos

x_values = [0.0, 0.5, 1.0, 1.5, 2.0,
            2.5, 3.0, 3.5, 4.0, 4.5]

cos_values = [cos(0.0), cos(0.5), cos(1.0), cos(1.5), cos(2.0),
              cos(2.5), cos(3.0), cos(3.5), cos(4.0), cos(4.5)]

plt.plot(x_values, cos_values, marker="o")
plt.axhline(0, color="black")
plt.title("Cosine Function")
plt.xlabel("x")
plt.ylabel("cos(x)")
plt.grid()
plt.show()
```

The graph resembles the cosine function and crosses zero between x  1.5 and x = 2.0, approximately at x = 1.57
