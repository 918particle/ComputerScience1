# Exercise 1: Evaluating a Function

```python
from math import cos
```

```python
x = 0.0, 0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5
```

```python
print(0.0, cos(0.0))
print(0.5, cos(0.5))
print(1.0, cos(1.0))
print(1.5, cos(1.5))
print(2.0, cos(2.0))
print(2.5, cos(2.5))
print(3.0, cos(3.0))
print(3.5, cos(3.5))
print(4.0, cos(4.0))
print(4.5, cos(4.5))
```

```python
import matplotlib.pyplot as plt
y = [cos(x) for x in x]
plt.plot(x, y, marker='o', color='b', linestyle='-')
plt.title("Graph")
plt.xlabel("X Axis Label")
plt.ylabel("Y Axis Label")
plt.show()
```
