```python
print("Multiplication Table")
print()
row = [1, 2, 3, 4, 5]
for i in range (6):
  if i ==1:
    print(" ",i, str(row))
  else:
    multiplied_row = [num * i for num in row]
    print(" ", i, multiplied_row)








```
