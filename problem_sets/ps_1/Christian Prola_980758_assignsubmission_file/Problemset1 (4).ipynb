```python
# Exercise 1: Evaluating a Function.

# Import the cosine function from the math library:
from math import cos

# Print x and cos(x) for ten different x-values:
print(0.0, cos(0.0))
print(0.5, cos(0.5))
print(1.0, cos(1.0))
print(1.5, cos(1.5))
print(2.0, cos(2.0))
print(2.5, cos(2.5))
print(3.0, cos(3.0))
print(3.5, cos(3.5))
print(4.0, cos(4.0))
print(4.5, cos(4.5))
```

```python
# Exercise 3: Class Rosters.
# class_rosters.py.

# Create a list of students for Computer Science 1:
cs_students = [
    "Ada Lovelace",
    "Alan Turing",
    "Grace Hopper"
]

# Create a list of students for Calculus I:
calculus_students = [
    "Katherine Johnson",
    "Isaac Newton",
    "Emmy Noether"
]

# Create a list of students for General Physics I:
physics_students = [
    "Marie Curie",
    "Albert Einstein",
    "Richard Feynman"
]

# Create a roster for each course:
cs_roster = [
    "Computer Science 1",
    cs_students
]

calculus_roster = [
    "Calculus I",
    calculus_students
]

physics_roster = [
    "General Physics I",
    physics_students
]

# Combine the three class rosters into one tuple:
class_rosters = (
    cs_roster,
    calculus_roster,
    physics_roster
)

# Print the completed class roster structure:
print(class_rosters)

# Nested structure of class_rosters:
# (
#     ["Computer Science 1", [student, student, student]],
#     ["Calculus I", [student, student, student]],
#     ["General Physics I", [student, student, student]]
# )
#
# ( ) = tuple
# [ ] = list
```
