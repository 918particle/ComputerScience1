```python
name = input("Student name:")

subjects_for_grades = [
    'Computer Science 1',
    'Calculus I',
    'General Physics I',
    'English Composition',
    'First-Year Seminar'
]

grades_list = [
    input(f'{subjects_for_grades[0]}: '),
    input(f'{subjects_for_grades[1]}: '),
    input(f'{subjects_for_grades[2]}: '),
    input(f'{subjects_for_grades[3]}: '),
    input(f'{subjects_for_grades[4]}: ')
]

print(f"\nReport Card \nStudent: {name} \n")

print(f'{subjects_for_grades[0]}...{grades_list[0]}')
print(f'{subjects_for_grades[1]}...........{grades_list[1]}')
print(f'{subjects_for_grades[2]}....{grades_list[2]}')
print(f'{subjects_for_grades[3]}..{grades_list[3]}')
print(f'{subjects_for_grades[4]}...{grades_list[4]}')
```
