```python
def placement(score):
 if score <= 1:
  return "Algebra"
 elif score <= 3:
  return "Trigonometry"
 else:
    return "Calculus"

students = [
("Maya", 20681001, 5),
("Noah", 20681002, 1),
("Olivia", 20681003, 3),
("Liam", 20681004, 4),
("Sophia", 20681005, 2),
("Ethan", 20681006, 0)
]

algebra = ["Algebra", []]
trig = ["Trigonometry", []]
calculus = ["Calculus", []]
math_courses = (algebra, trig, calculus)
```

```python
for student in students:
    name, student_id, score = student

    course = placement(score)

    if course == "Algebra":
        algebra[1].append((name, student_id))
    elif course == "Trigonometry":
        trig[1].append((name, student_id))
    else:
        calculus[1].append((name, student_id))

print(math_courses)
```

```python
i = 0
total = 0

while i < len(math_courses):
    total += len(math_courses[i][1])
    i += 1

print("Total students:", total)
```
