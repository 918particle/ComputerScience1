```python
def f(x):
  return (x+1)/(x-1)
values = [0, 0.5, 0.9,1.1, 1.5, 2]
values.append(0.99)
values.append(1.01)
for x in values:
  print(x,f(x))
```

x = 1 is not incuded in these values because the denominator would be zero.

The magnitude becomes extremly large because as x gets close to 1 from the left it becomes a large negative number and as x gts close from the right, it becomes a very large positve number

```python
def placement(score):
    if score >=4:
        return "Calculus"
    elif score >=3:
        return "Trigonomety"
    else:
        return "Algebra"
print(placement(1))
print(placement(3))
print(placement(5))

students = [("Maya", 20681001, 5),("Noah", 20681002, 1), ("Olivia", 20681003, 3),("Liam", 20681004, 4),("Sophia", 20681005, 2),("Ethan", 20681006, 0)]
algebra = ["Algebra", []]
trig = ["Trigonometry", []]
calculus = ["Calculus", []]
math_courses = (algebra, trig, calculus)

for student in students:
  name = student [0]
  student_id = student[1]
  score = student[2]
  course = placement(score)
  if course == "Algebra":
    algebra[1].append((name, student_id))
  elif course == "Trigonomety":
    trig[1].append((name, student_id))
  else:
    calculus[1].append((name, student_id))

print(math_courses)
```

```python
i = 0
total = 0

while i < len(math_courses):
    total += len(math_courses[i][1])
    i += 1
print(total)
```
