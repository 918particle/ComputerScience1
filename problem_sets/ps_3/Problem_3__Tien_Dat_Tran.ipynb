```python
def f(x):
    return (x + 1) / (x - 1)

values = [0, 0.5, 0.9, 1.1, 1.5, 2]

for x in values:
    print(x, f(x))
```

```python
def f(x):
    return (x + 1) / (x - 1)

values = [0, 0.5, 0.9, 0.99, 1.01, 1.1, 1.5, 2]

for x in values:
    print(x, f(x))
```

```python
def placement(score):
    if score <= 1:
        return "Algebra"
    elif score <= 3:
        return "Trigonometry"
    else:
        return "Calculus"


print(placement(1))
print(placement(3))
print(placement(5))


students = [
    ("Maya", 20681001, 5),
    ("Noah", 20681002, 1),
    ("Olivia", 20681003, 3),
    ("Liam", 20681004, 4),
    ("Sophia", 20681005, 2),
    ("Ethan", 20681006, 0)
]

algebra = ["Algebra", []]
trig = ["Trigonometry", []]
calculus = ["Calculus", []]

math_courses = (algebra, trig, calculus)

for student in students:
    name, student_id, score = student

    course = placement(score)

    if course == "Algebra":
        algebra[1].append((name, student_id))
    elif course == "Trigonometry":
        trig[1].append((name, student_id))
    else:
        calculus[1].append((name, student_id))

print(math_courses)
```
