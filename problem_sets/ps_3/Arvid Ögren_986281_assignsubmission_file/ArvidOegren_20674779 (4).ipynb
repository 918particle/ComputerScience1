```python
def placement(score):
 if score >= 4:
  return "Calculus"
 elif score >= 2:
    return "Trigonometry"
 else:
    return "Algebra"

print(placement(1))
print(placement(3))
print(placement(5))

students = [("Maya", 20681001, 5),
("Noah", 20681002, 1),
("Olivia", 20681003, 3),
("Liam", 20681004, 4),
("Sophia", 20681005, 2),
("Ethan", 20681006, 0)]

algebra = ["Algebra", []]
trig = ["Trigonometry", []]
calculus = ["Calculus", []]
math_courses = (algebra, trig, calculus)


for student in students:
  name, student_id, score = student
  student_tuple = (name, student_id)
  if placement(score) == "Calculus":
    calculus[1].append(student_tuple)
  elif placement(score) == "Trigonometry":
   trig[1].append(student_tuple)
  else:
   algebra[1].append(student_tuple)


print(math_courses)
```

```python
i = 0
j = 0
r = 0
total = 0

while i < len(math_courses[0][1]):
 math_courses[i]
 total += 1
 i += 1

while j < len(math_courses[1][1]):
 math_courses[j]
 total += 1
 j += 1

 while r < len(math_courses[2][1]):
  math_courses[r]
  total += 1
  r += 1

print(total)

# 6, since it's 6 students in total, across the courses.
```
