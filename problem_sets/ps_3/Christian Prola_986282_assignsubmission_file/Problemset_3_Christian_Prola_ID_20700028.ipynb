```python
# Exercise 1: Evaluating a Function

# Define the mathematical function
def f(x):
    return (x + 1) / (x - 1)

# Values to test
values = [0, 0.5, 0.9, 0.99, 1.01, 1.1, 1.5, 2]

# Evaluate and print x and f(x) for every value
for x in values:
    print(x, f(x))
```

x = 1 is not included because the denominator becomes zero, so the function is undefined.

Near x = 1, the magnitude of f(x) becomes very large. The value is negative just below 1 and positive just above 1.

```python
# Exercise 2: Placement Function

# Return the correct course based on the placement score
def placement(score):
    if score <= 1:
        return "Algebra"
    elif score <= 3:
        return "Trigonometry"
    else:
        return "Calculus"

# Test the placement function
print(placement(1))
print(placement(3))
print(placement(5))

# Student information
students = [
    ("Maya", 20681001, 5),
    ("Noah", 20681002, 1),
    ("Olivia", 20681003, 3),
    ("Liam", 20681004, 4),
    ("Sophia", 20681005, 2),
    ("Ethan", 20681006, 0)
]

# Create empty course rosters
algebra = ["Algebra", []]
trig = ["Trigonometry", []]
calculus = ["Calculus", []]

# Combine the course rosters into one tuple
math_courses = (
    algebra,
    trig,
    calculus
)

# Process every student
for student in students:
    name = student[0]
    student_id = student[1]
    score = student[2]

    # Use the placement function to determine the course
    course = placement(score)

    # Add the student's name and ID to the correct roster
    if course == "Algebra":
        algebra[1].append((name, student_id))
    elif course == "Trigonometry":
        trig[1].append((name, student_id))
    else:
        calculus[1].append((name, student_id))

# Print the completed course rosters
print(math_courses)
```

```python
# Exercise 3: While-Loop Count

# Start at the first course
i = 0

# Start the total student count at zero
total = 0

# Count the students in each course roster
while i < 3:
    total = total + len(math_courses[i][1])
    i = i + 1

# Print the total number of students
print(total)
```

The program should produce 6 because there are six students total, and each student appears exactly once in one of the three course rosters.
