```python
def f(x):
    return (x + 1) / (x - 1)

values = [0, 0.5, 0.9, 0.99, 1.01, 1.1, 1.5, 2]

for x in values:
    print(x, f(x))
```

```python
def placement(score):
    if score <= 1:
        return "Algebra"
    elif score <= 3:
        return "Trigonometry"
    else:
        return "Calculus"


print(placement(1))
print(placement(3))
print(placement(5))


students = [
    ("Maya", 20681001, 5),
    ("Noah", 20681002, 1),
    ("Olivia", 20681003, 3),
    ("Liam", 20681004, 4),
    ("Sophia", 20681005, 2),
    ("Ethan", 20681006, 0)
]

algebra = ["Algebra", []]
trig = ["Trigonometry", []]
calculus = ["Calculus", []]

math_courses = (
    algebra,
    trig,
    calculus
)

for student in students:
    name = student[0]
    student_id = student[1]
    score = student[2]

    course = placement(score)

    if course == "Algebra":
        algebra[1].append((name, student_id))
    elif course == "Trigonometry":
        trig[1].append((name, student_id))
    else:
        calculus[1].append((name, student_id))

print(math_courses)
```

```python
i = 0
total = 0

while i < 3:
    total = total + len(math_courses[i][1])
    i = i + 1

print(total)
```

x = 1 is not included because the denominator becomes 0, which would cause division by zero.

Near x = 1, the magnitude of f(x) becomes larger than normal. So 0.99 the value is negative, and at 1.01 the value is positive.
