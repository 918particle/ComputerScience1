```python
# exerise 2: Placement  function

# Return the correct course based on gthe placement score
def placement(score):
   if score <=1:
      return "Algebra"
   elif score <=3:
    return "Trigonometry"
   else:
    return "calculas"
# test the placement function
print(placement(1))
print(placement(3))
print(placement(5))

# Students information
students = [
    ("Maya", 20681001, 5),
    ("noah", 20681002, 1),
    ("Olivia", 20681003, 3),
    ("Liam", 20681004, 4),
    ("Sophia", 20681005, 2),
    ("Ethan", 20681006, 0),
]

# create empty course rosters
algebra = ["algebra", []]
trig = ["Trigonometry", []]
calculas = ["calulcas", []]

#combine the course rosters into tople
math_courses = [
    algebra,
    trig,
    calculas,
]

# processs everyh student
for student in students:
  name = student[0]
  student_id = student[1]
  score = student[2]

  #Use the placement function to determine the course
  course= placement(score)

  # add the students name and ID to the correct roster
  if course == "Algebra":
    algebra[1]. append((name, student_id))
  elif course =="Trigonometry":
    trig[1].append((name, student_id))
  else:
    calculas[1].append((name, student_id))

# print the completed course rosters
print(math_courses)
```

```python
# Exercise 3:

#start at the first course
i=0

#Start the total studnet count at zero
total = 0

# Count the students in each course roster
while i<3:
  total = total + len(math_courses[i][1])
  i = i + 1
# Print the total number of students
print(total)
```
