```python
def placement(score):
    if score >= 4:
        return "calc"
    elif score >= 2:
        return "trig"
    else:
        return "algebra"

students = [
    ("Maya", 20681001, 5),
    ("Noah", 20681002, 1),
    ("Olivia", 20681003, 3),
    ("Liam", 20681004, 4),
    ("Sophia", 20681005, 2),
    ("Ethan", 20681006, 0)
]

algebra = ["Algebra", []]
trig = ["Trigonometry", []]
calculus = ["Calculus", []]
math_courses = (algebra, trig, calculus)

for name, student_id, score in students:
    course_assignment = placement(score)
    if course_assignment == "calc":
        calculus[1].append(name)
    elif course_assignment == "trig":
        trig[1].append(name)
    else:
        algebra[1].append(name)


student_scores_map = {s_name: s_score for s_name, _, s_score in students}

print("\n--- Roster ---")
for course in math_courses:
    student_names_str = ""
    if course[1]:
        idx = 0
        for student_name in course[1]:
            score = student_scores_map[student_name]
            formatted_name = f"{student_name} ({score})"
            if idx == 0:
                student_names_str += formatted_name
            else:
                student_names_str += ", " + formatted_name
            idx += 1
    else:
        student_names_str = "No students"
    print(f"{course[0]}: {student_names_str}")
```

```python
num_math_courses = 0
for _ in math_courses:
    num_math_courses += 1

i = 0
total = 0
while i < num_math_courses:
    course = math_courses[i]
    student_count_in_course = 0
    for _ in course[1]:
        student_count_in_course += 1
    total += student_count_in_course
    i += 1

print(f"Total number of students across all courses: {total}")
```
