```python
def f(x):
    return (x + 1) / (x - 1)

values = [0, 0.5, 0.9, 1.1, 1.5, 2]

for x in values:
    print(x, f(x))

values = [0.99, 1.01]

for x in values:
    print(x, f(x))



def placement(score):
    if score <= 1:
        return "Algebra"
    elif score <= 3:
        return "Trigonometry"
    else:
        return "Calculus"


print(placement(1))
print(placement(3))
print(placement(5))


students = [
    ("Meya", 20681001, 5),
    ("Noah", 20681002, 1),
    ("Olivia", 20681003, 3),
    ("Liam", 20681004, 4),
    ("Sophia", 20681005, 2),
    ("Ethan", 20681006, 0)
]

algebra = []
trig = []
calculus = []

for student in students:
    name = student[0]
    student_id = student[1]
    score = student[2]

    course = placement(score)

    if course == "Algebra":
        algebra.append(student)
    elif course == "Trigonometry":
        trig.append(student)
    else:
        calculus.append(student)


print("Algebra:", algebra)
print("Trigonometry:", trig)
print("Calculus:", calculus)



total = 0

i = 0

while i < len(algebra):
    total = total + 1
    i = i + 1

i = 0

while i < len(trig):
    total = total + 1
    i = i + 1

i = 0

while i < len(calculus):
    total = total + 1
    i = i + 1

print("Total:", total)


```
