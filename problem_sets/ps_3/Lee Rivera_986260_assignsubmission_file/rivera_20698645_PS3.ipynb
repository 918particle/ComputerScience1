## Exercise 1: Evaluating a Function

```python
def f(x):
  return (x + 1) / (x - 1)
```

```python
values = [0, 0.5, 0.9,
          1.1, 1.5, 2]
```

```python
for x in values:
  print(x, f(x))
```

x = 1 is not included in the values as it would result in 2 / 0 and returns in an error -- this is due to numbers being unable to be divisible by 0.

However, a rational function's pole happens when the denominator (and not including the numerator) equals zero, which makes x = 1 the true pole.

```python
values.append(0.99)
values.append(1.01)

for x in values:
  print(x, f(x))
```

As f(x) nears the pole (1), the magnitude approaches infinity as the denominator gets closer to 0 simultaneously.

The sign of f(x) depends largely on which side 1 (the pole) is approached. From the left, with values lesser than 1, the sign is negative. However, on the right with values greater than 1, the sign is positive.

## Exercise 2: Placement Function

```python
def placement(score):
  if score >= 4:
    calculus[1].append(name, student_id)
    return "Calculus"
  elif score >= 2:
    trig[1].append(name, student_id)
    return "Trigonometry"
  else:
    algebra[1].append(name, student_id)
    return "Algebra"
```

```python
print(placement(1))
print(placement(3))
print(placement(5))
```

```python
students = [
  ("Maya", 20681001, 5),
  ("Noah", 20681002, 1),
  ("Olivia", 20681003, 3),
  ("Liam", 20681004, 4),
  ("Sophia", 20681005, 2),
  ("Ethan", 20681006, 0)
]
```

```python
algebra = ["Algebra", []]
trig = ["Trigonometry", []]
calculus = ["Calculus", []]
math_courses = (algebra, trig, calculus)
```

```python
for student in students:
  name = student[0]
  student_id = student[1]
  score = student[2]
  placement(score)
```

```python
print(math_courses)
```
