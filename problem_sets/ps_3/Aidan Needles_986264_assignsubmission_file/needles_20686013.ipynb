```python
def placement(score):
  if score >=4:
     return "Calculus"
  elif score>=2:
     return "Trignonmetry"
  else:
     return "Algebra"
students=[("Maya", 20681001, 5),
          ("Noah", 20681002, 1),
          ("Olivia", 20681003, 3),
          ("Liam", 20681004, 4),
          ("Sophia", 20681005, 2),
          ("Ethan", 20681006, 0)
          ]
algebra=["Algebra",[]]
trig=["Trigonometry", []]
calculus=["Calculus", []]
math_courses=(algebra,trig,calculus)

for name, student_id, score in students:
  course=placement(score)
  if course == "Algebra":
    algebra[1].append((name, student_id))
  elif course == "Trignonmetry":
    trig[1].append((name, student_id))
  elif course == "Calculus":
    calculus[1].append((name, student_id))
print(math_courses)


i = 0
total = 0
rosters = list(math_courses)

while i < len(rosters):
  total += len(rosters[i])
  i += 1

print(total)
```
