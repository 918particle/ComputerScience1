```python
def f(x):
  return (x + 1) / (x - 1)

values = [0, 0.5, 0.9, 1.1, 1.5, 2, 0.99, 1.01]

for i in values:
  print(f(i))
```

x = 1 is not included because nothing can divided by 0
What happens to the magnitude and sign of f (x) near the pole? -> Magnitude increases a lot, and the right side will be positive and the left side will be negative

```python
def placement(score):
  if score == 0 or score == 1:
    return "Algebra"
  elif score == 2 or score == 3:
    return "Trigonometry"
  else:
    return "Calculus"

print(placement(1))
print(placement(3))
print(placement(5))

students = [
("Maya", 20681001, 5),
("Noah", 20681002, 1),
("Olivia", 20681003, 3),
("Liam", 20681004, 4),
("Sophia", 20681005, 2),
("Ethan", 20681006, 0)
]

algebra = ["Algebra", []]
trig = ["Trigonometry", []]
calculus = ["Calculus", []]
math_courses = (algebra, trig, calculus)

for i in students:
  name = i[0]
  id = i[1]
  score = i[2]
  if placement(score) == "Algebra":
    algebra[1].append(i[0])
  elif placement(score) == "Trigonometry":
    trig[1].append(i[0])
  else:
    calculus[1].append(i[0])

print(algebra)
print(trig)
print(calculus)
print(math_courses)

i = 0
total = 0
while i < 3:
  total += len(math_courses[i])
  i += 1

print(total)
```
