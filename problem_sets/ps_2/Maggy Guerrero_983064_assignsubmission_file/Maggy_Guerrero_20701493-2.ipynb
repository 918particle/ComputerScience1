```python
#2: Sorting with a While loop

algebra2 = ["Algebra", []]
trigonometry2 = ["Trigonometry", []]
calculas2 = ["Calculas", []]

math_courses2 = (
    algebra2,
    trigonometry2,
    calculas2
)

i = 0

while i < 12:
  name = students[i][0]
  score = students[i][1]

  if score <= 1:
    algebra2[1].append(name)
  elif score <= 3:
    trigonometry2[1].append(name)
  else:
    calculas2[1].append(name)

  i = i +1
print(math_courses2)
```
