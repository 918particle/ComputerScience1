```python
# Exercise 1: Sorting with a For Loop

# List of students and their placement test scores
students = [
    ("Aaliyah", 4),
    ("Ben", 1),
    ("Camila", 3),
    ("Diego", 5),
    ("Elena", 2),
    ("Felix", 0),
    ("Grace", 3),
    ("Hector", 4),
    ("Isabel", 1),
    ("Jamal", 5),
    ("Keira", 0),
    ("Luis", 2)
]

# Create the three course rosters
algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]

# Combine the course rosters into one tuple
math_courses = (
    algebra,
    trigonometry,
    calculus
)

# Check each student and place them in the correct course
for student in students:
    name = student[0]
    score = student[1]

    if score <= 1:
        algebra[1].append(name)
    elif score <= 3:
        trigonometry[1].append(name)
    else:
        calculus[1].append(name)

# Print all course rosters together
print(math_courses)

# Print each course roster separately
print(math_courses[0])
print(math_courses[1])
print(math_courses[2])
```

```python
# Exercise 2: Sorting with a While Loop

# Original student list for Exercise 1 and 2
students = [
    ("Aaliyah", 4),
    ("Ben", 1),
    ("Camila", 3),
    ("Diego", 5),
    ("Elena", 2),
    ("Felix", 0),
    ("Grace", 3),
    ("Hector", 4),
    ("Isabel", 1),
    ("Jamal", 5),
    ("Keira", 0),
    ("Luis", 2)
]

# Create new empty course rosters
algebra2 = ["Algebra", []]
trigonometry2 = ["Trigonometry", []]
calculus2 = ["Calculus", []]

# Combine the course rosters into one tuple
math_courses2 = (
    algebra2,
    trigonometry2,
    calculus2
)

# Start with the first student in the list
i = 0

# Go through all 12 students
while i < 12:
    name = students[i][0]
    score = students[i][1]

    # Place the student in the correct course
    if score <= 1:
        algebra2[1].append(name)
    elif score <= 3:
        trigonometry2[1].append(name)
    else:
        calculus2[1].append(name)

    # Move to the next student
    i = i + 1

# Print all course rosters
print(math_courses2)

# Explanation:
# i = i + 1 moves the loop to the next student.
# If it is omitted, i never changes and the while loop continues forever.

# Comparison:
# In the for loop, student represents the current student.
# In the while loop, i determines which student is selected from the list.
```

```python
# Exercise 3: Math Placement System
# math_placement.py

# Create an empty list for the students
students = []

# Collect information for exactly 10 students
i = 0

while i < 10:
    name = input("Student name: ")
    score = int(input("Math score: "))

    # Store the student as a tuple
    student = (name, score)

    # Add the student tuple to the students list
    students.append(student)

    # Move to the next student
    i = i + 1


# Create the three empty course rosters
algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]

# Combine the course rosters into one tuple
math_courses = (
    algebra,
    trigonometry,
    calculus
)


# Sort each student into the correct course
for student in students:
    name = student[0]
    score = student[1]

    if score <= 1:
        algebra[1].append(name)
    elif score <= 3:
        trigonometry[1].append(name)
    else:
        calculus[1].append(name)


# Print the final math placement report
print()
print("Math Placement Report")

print(math_courses[0][0] + ":")
print(math_courses[0][1])

print(math_courses[1][0] + ":")
print(math_courses[1][1])

print(math_courses[2][0] + ":")
print(math_courses[2][1])
```
