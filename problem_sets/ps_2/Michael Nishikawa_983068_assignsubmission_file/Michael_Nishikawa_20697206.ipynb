```python
students = [
    ("Aaliyah", 4),
    ("Ben", 1),
    ("Camila", 3),
    ("Diego", 5),
    ("Elena", 2),
    ("Felix", 0),
    ("Grace", 3),
    ("Hector", 4),
    ("Isabel", 1),
    ("Jamal", 5),
    ("Keira", 0),
    ("Luis", 2)
]
algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]
math_courses = (
    algebra,
    trigonometry,
    calculus
)
for student in students:
    name = student[0]
    score = student[1]
    if score == 0 or score == 1:
        algebra[1].append(name)
    elif score == 2 or score == 3:
        trigonometry[1].append(name)
    else:
        calculus[1].append(name)
print(math_courses)
print(math_courses[0])
print(math_courses[1])
print(math_courses[2])
```

```python
students = []
i = 0
while i < 10:
    name = input("Student name: ")
    score = int(input("Math score: "))
    student = (name, score)
    students.append(student)
    i = i + 1
algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]
math_courses = (
    algebra,
    trigonometry,
    calculus
)
for student in students:
    name = student[0]
    score = student[1]
    if score >= 4:
        calculus[1].append(name)
    elif score >= 2:
        trigonometry[1].append(name)
    else:
        algebra[1].append(name)
print("Math Placement Report")
print(math_courses[0][0] + ":")
print(math_courses[0][1])
print(math_courses[1][0] + ":")
print(math_courses[1][1])
print(math_courses[2][0] + ":")
print(math_courses[2][1])
```
