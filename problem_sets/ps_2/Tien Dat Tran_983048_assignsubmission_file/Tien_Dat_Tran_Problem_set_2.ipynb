```python
students = [
("Aaliyah", 4),
("Ben", 1),
("Camila", 3),
("Diego", 5),
("Elena", 2),
("Felix", 0),
("Grace", 3),
("Hector", 4),
("Isabel", 1),
("Jamal", 5),
("Keira", 0),
("Luis", 2)
]
algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]

math_courses = (
algebra,
trigonometry,
calculus
)

for students in students:
  name = students[0]
  score = students [1]
  if score  == 0:
    algebra[1].append(name)
  elif score == 1:
    algebra[1].append(name)
  elif score == 2:
    trigonometry[1].append(name)
  elif score == 3:
    trigonometry[1].append(name)
  elif score == 4:
    calculus[1].append(name)
  else:
    calculus[1].append(name)

print(math_courses)
print(math_courses[0])
print(math_courses[1])
print(math_courses[2])
```

```python
students = [
("Aaliyah", 4),
("Ben", 1),
("Camila", 3),
("Diego", 5),
("Elena", 2),
("Felix", 0),
("Grace", 3),
("Hector", 4),
("Isabel", 1),
("Jamal", 5),
("Keira", 0),
("Luis", 2)
]

algebra2 = ["Algebra", []]
trigonometry2 = [
"Trigonometry", []
]
calculus2 = ["Calculus", []]

math_courses2 = (
algebra2,
trigonometry2,
calculus2
)

i = 0
while i < 12:
  name = students[i][0]
  score = students[i][1]
  if score == 0:
    algebra2[1].append(name)
  elif score == 1:
    algebra2[1].append(name)
  elif score == 2:
    trigonometry2[1].append(name)
  elif score == 3:
    trigonometry2[1].append(name)
  elif score == 4:
    calculus2[1].append(name)
  else:
    calculus2[1].append(name)
  i = i + 1

print(math_courses2)
print(math_courses2[0])
print(math_courses2[1])
print(math_courses2[2])
```

The i = i + 1 line is required, cause the line will moves the loop to the next student. If not, it only stay at 0 so creatng a never end loop.

With the "for" loop: the "student" holds the student tuple
"while" loop: the "i" will select the student in the list

```python
students = []

i = 0
while i < 10:
  name = input("Student name: ")
  score = int(input("Math score: "))
  student = (name, score)
  students.append(student)
  i = i + 1

algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]

math_courses = (algebra, trigonometry, calculus)

for i in students:
  name = i[0]
  score = i[1]
  if score == 0:
    algebra[1].append(name)
  elif score == 1:
    algebra[1].append(name)
  elif score == 2:
    trigonometry[1].append(name)
  elif score == 3:
    trigonometry[1].append(name)
  elif score == 4:
    calculus[1].append(name)
  else:
    calculus[1].append(name)

print("Math Placement Report")
print("Algebra:")
print(math_courses[0][1])
print("Trigonometry:")
print(math_courses[1][1])
print("Calculus:")
print(math_courses[2][1])
```
