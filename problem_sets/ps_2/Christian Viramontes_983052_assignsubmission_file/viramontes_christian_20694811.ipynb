```python
students = [
("Aaliyah", 4),
("Ben", 1),
("Camila", 3),
("Diego", 5),
("Elena", 2),
("Felix", 0),
("Grace", 3),
("Hector", 4),
("Isabel", 1),
("Jamal", 5),
("Keira", 0),
("Luis", 2)]

algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]

math_courses = (algebra, trigonometry, calculus)

for student in students:
    name = student[0]
    score = student[1]
# conditional logic goes here
    if score >= 4:
      calculus[1].append(name)

    elif score >= 2:
      trigonometry[1].append(name )

    else:
      algebra[1].append(name)

print(math_courses[0])
print(math_courses[1])
print(math_courses[2])
```

```python
students = [
("Aaliyah", 4),
("Ben", 1),
("Camila", 3),
("Diego", 5),
("Elena", 2),
("Felix", 0),
("Grace", 3),
("Hector", 4),
("Isabel", 1),
("Jamal", 5),
("Keira", 0),
("Luis", 2)
]

algebra2 = ["Algebra", []]
trigonometry2 = ["Trigonometry", []]
calculus2 = ["Calculus", []]

math_courses2 = (algebra2, trigonometry2, calculus2)

i = 0
while i < 12:
    name = students[i][0]
    score = students[i][1]
    # conditional logic goes here
    if score >= 4:
        calculus2[1].append(name)
    elif score >= 2:
        trigonometry2[1].append(name)
    else:
        algebra2[1].append(name)
    i = i + 1

print(math_courses2[0])
print(math_courses2[1])
print(math_courses2[2])
```

i = i + 1 must occur during every pass through the loop because it allows the loop to stop runing, with out it, the loop contiunously runs and google colab does not like it.

In the for loop, the variable student in students represnets the current student. In the while loop, the variable i < 12 and the if, elif, else statement determines which student is selected from the list
