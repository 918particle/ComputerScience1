```python
students = [("Aaliyah", 4),
("Ben", 1),
("Camila", 3),
("Diego", 5),
("Elena", 2),
("Felix", 0),
("Grace", 3),
("Hector", 4),
("Isabel", 1),
("Jamal", 5),
("Keira", 0),
("Luis", 2)]

algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]

math_courses = [(algebra), (trigonometry), (calculus)]

for student in students:
  name = student[0]
  score = student[1]

  if int((score)) >= 4:
    calculus[1].append(str(name))
  elif int((score)) > 2:
    trigonometry[1].append(str(name))
  else:
    algebra[1].append(str(name))

print(math_courses)

print(math_courses[0])
print(math_courses[1])
print(math_courses[2])


```

```python
students = [("Aaliyah", 4),
("Ben", 1),
("Camila", 3),
("Diego", 5),
("Elena", 2),
("Felix", 0),
("Grace", 3),
("Hector", 4),
("Isabel", 1),
("Jamal", 5),
("Keira", 0),
("Luis", 2)]

algebra2 = ["Algebra", []]
trigonometry2 = ["Trigonometry", []]
calculus2 = ["Calculus", []]

math_courses2 = [(algebra2), (trigonometry2), (calculus2)]

i = 0

while i < 12:
  name = students[i][0]
  score = students[i][1]
  i = i + 1

  if int((score)) >= 4:
    calculus2[1].append(str(name))
  elif int((score)) > 2:
    trigonometry2[1].append(str(name))
  else:
    algebra2[1].append(str(name))

# The statement "i = i + 1" must happen so each time the loop runs, the variable
# will become one unit larger so it will ultimely stop running at 12 (the amount
# of students). Othewise, the loop would have countinued in all infinity.


print(math_courses2)

print(math_courses2[0])
print(math_courses2[1])
print(math_courses2[2])

# In the for loop, the variable "student" represents the current student, while
# in the while loop the variable "students[i]" represents the current student
# evaulated.

```
