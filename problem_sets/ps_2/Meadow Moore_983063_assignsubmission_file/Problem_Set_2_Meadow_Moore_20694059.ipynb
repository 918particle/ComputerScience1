Exercise 1: Sorting with a For Loop

```python
students = [
("Aaliyah", 4),
("Ben", 1),
("Camila", 3),
("Diego", 5),
("Elena", 2),
("Felix", 0),
("Grace", 3),
("Hector", 4),
("Isabel", 1),
("Jamal", 5),
("Keira", 0),
("Luis", 2)
]
```

```python
algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]
math_courses = (algebra, trigonometry, calculus)
```

```python
for student in students:
    name = student[0]
    score = student[1]
    print(name, score)
```

```python
for student_name, score in students:
    if score >= 4:
        calculus[1].append(student_name)
    elif score >= 2:
        trigonometry[1].append(student_name)
    else:
        algebra[1].append(student_name)

print(algebra)
print(trigonometry)
print(calculus)
```
