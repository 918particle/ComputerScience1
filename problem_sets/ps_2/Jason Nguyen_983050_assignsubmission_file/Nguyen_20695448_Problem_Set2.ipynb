```python
students = []

algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]

math_courses = (
    algebra,
    trigonometry,
    calculus
)

i = 0

print("\n--- Course Summaries ---")

while i < 10:
    name = input('students name: ')
    score = int(input('math score: '))

    if score >= 4:
        algebra[1].append(name)
    elif score >= 3:
        trigonometry[1].append(name)
    else:
        calculus[1].append(name)
    i = i + 1

i = 0

while i < 3:
    course = math_courses[i]
    print(f"{course[0]}: ", end="")

    first_student = True

    for student_name in course[1]:
        if not first_student:
            print(", ", end="")
        print(student_name, end="")
        first_student = False

    print()
    i = i + 1
```
