```python
students = [
  ("Aaliyah", 4),
  ("Ben", 1),
  ("Camila", 3),
  ("Diego", 5),
  ("Elena", 2),
  ("Felix", 0),
  ("Grace", 3),
  ("Hector", 4),
  ("Isabel", 1),
  ("Jamal", 5),
  ("Keira", 0),
  ("Luis", 2)
]
```

## Exercise 1

```python
algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]

math_courses = (algebra, trigonometry, calculus)
```

```python
for student in students:
  name = student[0]
  score = student[1]
  if score >= 4:
    calculus[1].append(name)
  elif score >= 2:
    trigonometry[1].append(name)
  else:
    algebra[1].append(name)
```

```python
print(math_courses)
print()
print(math_courses[0])
print(math_courses[1])
print(math_courses[2])
```

## Exercise 2

```python
algebra2 = ["Algebra", []]
trigonometry2 = ["Trigonometry", []]
calculus2 = ["Calculus", []]

math_courses2 = (algebra2, trigonometry2, calculus2)
```

```python
i = 0

while i < 12:
  name = students[i][0]
  score = students[i][1]
  if score >= 4:
    calculus2[1].append(name)
  elif score >= 2:
    trigonometry2[1].append(name)
  else:
    algebra2[1].append(name)
  i = i + 1
```

Explain why

> i = i + 1

must occur during every pass through the loop. What would happen if this statement were omitted?

*If this statement was omitted, then only the 'first' item in the student list (Aaliyah) would be sorted in the while loop instead of all students incrementally sorted.*

```python
print(math_courses2)
```

In the for loop, what variable represents the current student?

*student*

In the while loop, what variable determines which student is selected from the list?

*i*

## Exercise 3

```python
students = []

i = 0
while i < 10:
  name = input("Student name: ")
  score = int(input("Math score: "))
# store this student
  student = (name, score)
  students.append(student)
  i = i + 1
```

```python
algebra = ["Algebra", []]
trigonometry = ["Trigonometry", []]
calculus = ["Calculus", []]

math_courses = (algebra, trigonometry, calculus)
```

```python
for student in students:
  name = student[0]
  score = student[1]

  if score >= 4:
    calculus[1].append(name)
  elif score >= 2:
    trigonometry[1].append(name)
  else:
    algebra[1].append(name)
```
